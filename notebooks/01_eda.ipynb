```python
df["LossRatio"] = np.where(
    df["TotalPremium"] > 0,
    df["TotalClaims"] / df["TotalPremium"],
    0
)

df["Margin"] = df["TotalPremium"] - df["TotalClaims"]
```

```python
df["CapitalOutstanding"] = pd.to_numeric(df["CapitalOutstanding"], errors="coerce")
```

```python
df["Cylinders"] = df["Cylinders"].replace(0, np.nan)
df["Cylinders"] = df["Cylinders"].fillna(df["Cylinders"].median())
```

```python
df["cubiccapacity"] = df["cubiccapacity"].fillna(df["cubiccapacity"].median())
df["kilowatts"] = df["kilowatts"].fillna(df["kilowatts"].median())
df["NumberOfDoors"] = df["NumberOfDoors"].fillna(df["NumberOfDoors"].median())
```

```python
cat_cols = ["VehicleType","make","Model","bodytype","VehicleIntroDate","mmcode"]

for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
df["Bank"] = df["Bank"].fillna("Unknown")
df["AccountType"] = df["AccountType"].fillna("Unknown")

df["Gender"] = df["Gender"].fillna(df["Gender"].mode()[0])
df["MaritalStatus"] = df["MaritalStatus"].fillna(df["MaritalStatus"].mode()[0])
```

```python
flag_cols = ["Converted","Rebuilt","WrittenOff","NewVehicle"]
df[flag_cols] = df[flag_cols].fillna("Unknown")
```

```python
df["CustomValueEstimate_missing"] = df["CustomValueEstimate"].isnull().astype(int)

df["CustomValueEstimate"] = df["CustomValueEstimate"].fillna(
    df["CustomValueEstimate"].median()
)
```

```python
import pandas as pd

df = pd.read_csv("../data/MachineLearningRating_v3.txt", sep="|")

df["LossRatio"] = df["TotalClaims"] / df["TotalPremium"].replace(0, 1)

df.groupby("Province")[["TotalPremium", "TotalClaims", "LossRatio"]].mean()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))
sns.barplot(data=df, x="Province", y="LossRatio")
plt.xticks(rotation=45)
plt.title("Loss Ratio by Province")
plt.show()
```

```python
df.groupby("VehicleType")[["TotalPremium", "TotalClaims", "LossRatio"]].mean().sort_values("LossRatio", ascending=False)
```

```python
df.groupby("Gender")[["TotalPremium", "TotalClaims", "LossRatio"]].mean()
```

Time trend

```python
df["TransactionMonth"] = pd.to_datetime(df["TransactionMonth"])
```

```python
monthly = df.groupby(df["TransactionMonth"].dt.to_period("M"))[["TotalPremium", "TotalClaims"]].mean()
monthly.plot(figsize=(12,5), title="Monthly Premium vs Claims Trend")
```
