```python
df['TotalPremium'] = pd.to_numeric(df['TotalPremium'], errors='coerce')
df['TotalClaims'] = pd.to_numeric(df['TotalClaims'], errors='coerce')
```

```python
df['LossRatio'] = df['TotalClaims'] / df['TotalPremium']
```
